```python
# 반복문에 포함된 조건문

number = int(input("숫자를 입력하세요"))

#총 합을 구하기 위해서는 반복문 전에 0으로 초기화 된 변수가 필요
x = 0
y = 0

for i in range(1, number+1):
    if i%2 == 0:
        print(i, "-->짝수")
        x += i
    else:
            print(i, "--> 홀수")
            y += i

print("짝수들의 합:", x)
print("홀수들의 합:", y)
```

### 리스트 원소를 직접 탐색하는 경우

```python
mydata = [[46,26,86,97], [55,21,98,33]]

size = len(mydata)
print("size =", size)

for i in range(0, size):
    print(i, "번째 데이터 ->", mydata[i])

# ------------------------------------------------------
    total = 0                 # 각 행 단위 합계이므로 매 행마다 합계 변수를 새로 초기화 한다           
    size = len(mydata[i])     # 각 행에 포함됨 리스트의 길이

    for j in range(0, size):  # mydata[i]에 대한 새로운 반복문 구성
        total += mydata[i][j]

    average = total / size    # 행 단위 평균값

    print(i, "번째 데이터 -> 총합 =", total, "평균=", average)
```
